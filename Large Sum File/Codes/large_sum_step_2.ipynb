```python
import pandas as pd
import os

# Function to modify a single CSV file
def modify_csv(csv_path, output_dir):
    # Read the CSV file into a DataFrame
    df = pd.read_csv(csv_path, encoding='latin-1')

    # Extract the necessary values for modification
    original_value = df.at[0, 'Running Number of Data File']
    temperature_measurement = df.at[0, 'Temperature of measurement']

    # Split the temperature measurement to get the values before and after the '-'
    split_temp = temperature_measurement.split('-')
    if len(split_temp) > 1:
        temp_before_dash = split_temp[0]
        temp_after_dash = split_temp[1]
    else:
        temp_before_dash = temperature_measurement  # Handle case where there's no '-'
        temp_after_dash = ''  # No value after the '-'

    # Modify the 'Running Number of Data File' column to append the value before '-'
    if temp_before_dash:
        df.at[0, 'Running Number of Data File'] = f"{original_value}-{temp_before_dash}"

    # Update the 'Temperature of measurement' column to only contain the value after the '-'
    df.at[0, 'Temperature of measurement'] = temp_after_dash

    # Create output file path
    file_name = os.path.basename(csv_path)
    output_csv_path = os.path.join(output_dir, f"modified_{file_name}")

    # Save the modified DataFrame to a new CSV file
    df.to_csv(output_csv_path, index=False, encoding='latin-1')

    print(f"Modified CSV file has been saved successfully: {output_csv_path}")

# Path to the directory containing the CSV files
input_dir = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\Large Sum File\Sum\Sum\CSV_Output_Step_1'
# Path to the output directory where modified CSV files will be saved
output_dir = r'C:\Users\Master.WR\OneDrive - stud.tu-darmstadt.de\Documents\Large Sum File\Sum\Sum\CSV_Output_Step_2'

# Ensure the output directory exists
os.makedirs(output_dir, exist_ok=True)

# Iterate over all CSV files in the input directory and apply the modification
for file_name in os.listdir(input_dir):
    if file_name.endswith('.csv'):
        csv_path = os.path.join(input_dir, file_name)
        modify_csv(csv_path, output_dir)
```
